```python
# Generar una secuencia de Fibonacci
a, b = 0, 1
while b < 100:
   print(b)
   a, b = b, a + b
```

```python
numero = 2
while numero < 100:
    divisor = 2
    while numero % divisor != 0:
        divisor += 1
    if divisor == numero:
        print(numero, end=" ")
    numero += 1
```

```python
n = int(input("Ingresa un número: "))
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print("factorial de", n, "es", factorial)
```
